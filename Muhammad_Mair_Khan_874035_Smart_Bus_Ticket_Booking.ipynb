# Smart Bus Ticket Booking System


```python
routes = [
    {"id": 101, "from": "Karachi", "to": "Hyderabad", "time": "08:00 AM", "fare": 1200, "seats": 40},
    {"id": 102, "from": "Karachi", "to": "Sukkur", "time": "09:30 AM", "fare": 2000, "seats": 36},
    {"id": 103, "from": "Hyderabad", "to": "Mirpur Khas", "time": "10:00 AM", "fare": 800, "seats": 32},
    {"id": 104, "from": "Karachi", "to": "Lahore", "time": "07:00 PM", "fare": 5000, "seats": 45},
    {"id": 105, "from": "Sukkur", "to": "Hyderabad", "time": "01:00 PM", "fare": 1500, "seats": 30}
]

cities = ("Karachi", "Hyderabad", "Sukkur", "Mirpur Khas", "Lahore")
bookings = []
used_ids = set()
next_id = 1001

print("Bus routes loaded:", len(routes))
```

```python
def find_route(route_id):
    for route in routes:
        if route["id"] == route_id:
            return route
    return None


def view_routes():
    print("\nAvailable Routes")
    for route in routes:
        print("Route ID:", route["id"])
        print("From:", route["from"], "To:", route["to"])
        print("Departure:", route["time"])
        print("Fare: Rs.", route["fare"])
        print("Seats available:", route["seats"])
        print("------------------------")


def search_route():
    start = input("Departure city: ").strip().title()
    destination = input("Destination city: ").strip().title()

    for route in routes:
        if route["from"] == start and route["to"] == destination:
            print("Route found:", route["id"])
            print("Departure:", route["time"])
            print("Fare: Rs.", route["fare"])
            print("Seats available:", route["seats"])
            break
    else:
        print("No route found.")
```

```python
def book_ticket():
    global next_id

    view_routes()

    try:
        route_id = int(input("Enter route ID: "))
    except ValueError:
        print("Enter a valid number.")
        return

    route = find_route(route_id)
    if route is None:
        print("Route not found.")
        return

    name = input("Passenger name: ").strip()
    if name == "":
        print("Name cannot be empty.")
        return

    try:
        age = int(input("Passenger age: "))
        seat_count = int(input("Number of seats: "))
    except ValueError:
        print("Age and seats must be numbers.")
        return

    if age <= 0 or seat_count <= 0:
        print("Age and seats must be greater than zero.")
        return

    if seat_count > route["seats"]:
        print("Not enough seats available.")
        return

    while next_id in used_ids:
        next_id += 1

    booking = {
        "id": next_id,
        "name": name,
        "age": age,
        "route_id": route["id"],
        "from": route["from"],
        "to": route["to"],
        "seats": seat_count,
        "fare": route["fare"] * seat_count,
        "cancelled": False
    }

    bookings.append(booking)
    used_ids.add(next_id)
    route["seats"] -= seat_count

    print("Ticket booked.")
    print("Booking ID:", next_id)
    print("Total fare: Rs.", booking["fare"])
    next_id += 1
```

```python
def view_bookings():
    if len(bookings) == 0:
        print("No bookings yet.")
        return

    for booking in bookings:
        print("\nBooking ID:", booking["id"])
        print("Passenger:", booking["name"])
        print("Route:", booking["from"], "to", booking["to"])
        print("Seats:", booking["seats"])
        print("Fare: Rs.", booking["fare"])
        if booking["cancelled"]:
            print("Status: Cancelled")
        else:
            print("Status: Confirmed")


def cancel_ticket():
    try:
        booking_id = int(input("Enter booking ID: "))
    except ValueError:
        print("Enter a valid booking ID.")
        return

    for booking in bookings:
        if booking["id"] == booking_id:
            if booking["cancelled"]:
                print("This booking is already cancelled.")
                return

            booking["cancelled"] = True
            route = find_route(booking["route_id"])
            if route is not None:
                route["seats"] += booking["seats"]

            print("Ticket cancelled.")
            return
    else:
        print("Booking ID not found.")
```

```python
def check_seats():
    print("\nSeat Availability")
    for route in routes:
        print(route["id"], route["from"], "to", route["to"],
              "- Seats:", route["seats"])


def show_schedule():
    print("\nBus Schedule")
    for city in cities:
        print("\nFrom", city)
        for route in routes:
            if route["from"] == city:
                print(route["time"], "-", route["to"], "(Route", route["id"], ")")


def show_report():
    total_bookings = 0
    total_income = 0

    for booking in bookings:
        if booking["cancelled"]:
            continue
        total_bookings += 1
        total_income += booking["fare"]

    print("\nBooking Report")
    print("Confirmed bookings:", total_bookings)
    print("Total income: Rs.", total_income)

    for number, booking in enumerate(bookings, start=1):
        if not booking["cancelled"]:
            print(number, booking["name"])
```

```python
def main():
    while True:
        print("\nSMART BUS TICKET BOOKING")
        print("1. View routes")
        print("2. Search route")
        print("3. Book ticket")
        print("4. View bookings")
        print("5. Cancel ticket")
        print("6. Check seats")
        print("7. View schedule")
        print("8. Booking report")
        print("9. Exit")

        choice = input("Choose an option: ")

        if choice == "1":
            view_routes()
        elif choice == "2":
            search_route()
        elif choice == "3":
            book_ticket()
        elif choice == "4":
            view_bookings()
        elif choice == "5":
            cancel_ticket()
        elif choice == "6":
            check_seats()
        elif choice == "7":
            show_schedule()
        elif choice == "8":
            show_report()
        elif choice == "9":
            print("Goodbye.")
            break
        else:
        
            print("Please choose a number from 1 to 9.")


main()
```
